# 1 项目背景与核心问题

## 1.1 工程背景
在理想流体力学中，圆柱绕流是平面势流的经典问题。单纯均匀来流复势（$U_0$ 为来流速度）无法满足**流体不能穿透圆柱表面（法向速度为零）**的物理边界条件。

本阶段目标是：通过引入偶极子项，构造一个既符合物理边界条件、又保持解析性的复势函数，为后续流场可视化与工程分析奠定数学基础。

## 1.2 基本假设
本推导基于理想流体平面势流假设：

- 不可压缩流体：密度 $\rho$ 为常数，连续性方程 $\nabla\cdot\boldsymbol{v}=0$；
- 无旋流动：速度场存在速度势函数 $\varphi$，满足 $\boldsymbol{v}=\nabla\varphi$；
- 平面流动：流动仅在 $x\text{-}y$ 平面内，所有物理量与 $z$ 坐标无关；
- 边界条件：圆柱表面 $r=a$ 处法向速度为 $0$，即流线贴合圆柱表面。

# 2 复势函数构造思路

## 2.1 基础流场：均匀来流
沿 $x$ 轴正方向的均匀来流复势为：
$$
F_1(z)=U_0 z
$$
其中 $z=x+iy$ 为复变量，$U_0$ 为来流速度。

速度势：
$$
\varphi_1=U_0 r\cos\theta
$$

流函数：
$$
\psi_1=U_0 r\sin\theta
$$

该流场的流线为平行于 $x$ 轴的直线，无法满足圆柱表面的闭合流线要求。

## 2.2 引入偶极子项
为满足圆柱边界的流线闭合条件，在均匀来流基础上引入偶极子复势：
$$
F_2(z)=\frac{M}{z}
$$
其中 $M$ 为偶极子强度，其物理本质为源-汇对极限（源与汇无限靠近且强度同步增大），可改变流场拓扑结构，使流线在圆柱表面闭合。

速度势：
$$
\varphi_2=\frac{M}{r}\cos\theta
$$

流函数：
$$
\psi_2=-\frac{M}{r}\sin\theta

# 3 边界条件匹配与偶极子强度推导

## 3.1 边界条件：圆柱表面法向速度为零
圆柱表面 $r=a$ 处，流体不可穿透，要求径向速度 $v_r=0$。

径向速度由速度势的偏导数定义：
$$
v_r=\frac{\partial\varphi}{\partial r}
$$

总复势为均匀来流与偶极子的叠加：
$$
F(z)=U_0 z+\frac{M}{z}
$$

对应总速度势：
$$
\varphi=U_0 r\cos\theta+\frac{M}{r}\cos\theta
$$

## 3.2 推导偶极子强度 $M$
对 $\varphi$ 求径向偏导，得到径向速度：
$$
v_r=\left(U_0-\frac{M}{r^2}\right)\cos\theta
$$

代入圆柱表面边界条件 $r=a$ 且 $v_r=0$：
$$
\left(U_0-\frac{M}{a^2}\right)\cos\theta=0
$$

对任意 $\theta$ 成立，因此系数必须满足：
$$
U_0-\frac{M}{a^2}=0
$$

解得偶极子强度：
$$
M=U_0 a^2

# 4 合成复势与极坐标展开

## 4.1 合成复势表达式
将 $M=U_0 a^2$ 代入偶极子复势，与均匀来流复势叠加，得到圆柱绕流总复势：
$$
F(z)=U_0\left(z+\frac{a^2}{z}\right)
$$
其中 $a$ 为圆柱半径，该复势在 $|z|\ge a$ 区域内解析，满足理想流体平面势流的数学要求。

## 4.2 极坐标形式展开
将 $z=re^{i\theta}$ 代入总复势，分离实部（速度势）与虚部（流函数）：

速度势：
$$
\varphi=U_0\left(r+\frac{a^2}{r}\right)\cos\theta
$$

流函数：
$$
\psi=U_0\left(r-\frac{a^2}{r}\right)\sin\theta
$$

## 4.3 边界条件验证
在圆柱表面 $r=a$ 处：
- 流函数：$\psi=0$，即圆柱表面为一条流线，满足流线闭合条件；
- 径向速度：$v_r=0$，满足无穿透边界条件。

# 5 极坐标柯西-黎曼（C-R）方程验证

## 5.1 极坐标C-R方程形式
解析复势需满足极坐标C-R方程：
$$
\begin{cases}
\dfrac{\partial\varphi}{\partial r}=\dfrac{1}{r}\dfrac{\partial\psi}{\partial\theta}\\[6pt]
\dfrac{1}{r}\dfrac{\partial\varphi}{\partial\theta}=-\dfrac{\partial\psi}{\partial r}
\end{cases}
$$

## 5.2 代入验证
第一式验证：
左侧：$\dfrac{\partial\varphi}{\partial r}=U_0\left(1-\dfrac{a^2}{r^2}\right)\cos\theta$  
右侧：$\dfrac{1}{r}\dfrac{\partial\psi}{\partial\theta}=U_0\left(1-\dfrac{a^2}{r^2}\right)\cos\theta$  
结论：左右两侧相等，第一式成立。

第二式验证：
左侧：$\dfrac{1}{r}\dfrac{\partial\varphi}{\partial\theta}=-U_0\left(1+\dfrac{a^2}{r^2}\right)\dfrac{\sin\theta}{r}$  
右侧：$-\dfrac{\partial\psi}{\partial r}=-U_0\left(1+\dfrac{a^2}{r^2}\right)\dfrac{\sin\theta}{r}$  
结论：左右两侧相等，第二式成立。

## 5.3 验证结论
圆柱绕流复势 $F(z)$ 在 $r\ge a$ 区域内满足极坐标C-R方程，证明其为解析函数，符合理想流体平面势流的数学要求。

# 6 复速度与物理场分析

## 6.1 复速度定义
复速度为复势的导数：
$$
\overline{V}=\frac{dF}{dz}=U_0\left(1-\frac{a^2}{z^2}\right)
$$

极坐标下速度分量：
径向速度：
$$
v_r=U_0\left(1-\frac{a^2}{r^2}\right)\cos\theta
$$

周向速度：
$$
v_\theta=-U_0\left(1+\frac{a^2}{r^2}\right)\sin\theta
$$

## 6.2 圆柱表面物理特性
在 $r=a$ 处：
- 径向速度 $v_r=0$，满足无穿透边界条件；
- 周向速度 $v_\theta=-2U_0\sin\theta$，合速度大小 $|V|=2U_0|\sin\theta|$；
- 压力系数（由伯努利方程）：$C_p=1-4\sin^2\theta$，体现圆柱表面压力分布规律。

# 7 数值验证与代码实现

## 7.1 符号验证（SymPy）
使用Sympy对复势、速度势、流函数、C–R 方程进行符号推导，验证复势解析性与边界条件。

```python
import sympy as sp

# 符号定义
r, theta, U0, a = sp.symbols('r theta U0 a', real=True, positive=True)

# 速度势与流函数
phi = U0 * (r + a**2 / r) * sp.cos(theta)
psi = U0 * (r - a**2 / r) * sp.sin(theta)

# C-R 方程验证
cr1 = sp.simplify(sp.diff(phi, r) - (1/r)*sp.diff(psi, theta))
cr2 = sp.simplify((1/r)*sp.diff(phi, theta) + sp.diff(psi, r))

print("C-R 方程 1 成立：", cr1 == 0)
print("C-R 方程 2 成立：", cr2 == 0)

# 边界 r=a 处径向速度
vr = sp.diff(phi, r)
vr_surf = sp.simplify(vr.subs(r, a))
print("r=a 处径向速度 v_r =", vr_surf)
```

## 7.2 数值计算验证

```python
import numpy as np

# 参数
U0 = 1.0
a = 1.0

# 测试点
r_test = 2.0
theta_test = np.pi / 4

# 计算
phi_test = U0 * (r_test + a**2 / r_test) * np.cos(theta_test)
psi_test = U0 * (r_test - a**2 / r_test) * np.sin(theta_test)
vr_test = U0 * (1 - a**2 / r_test**2) * np.cos(theta_test)
vt_test = -U0 * (1 + a**2 / r_test**2) * np.sin(theta_test)

print("速度势 φ =", round(phi_test, 4))
print("流函数 ψ =", round(psi_test, 4))
print("径向速度 v_r =", round(vr_test, 4))
print("周向速度 v_θ =", round(vt_test, 4))
```
