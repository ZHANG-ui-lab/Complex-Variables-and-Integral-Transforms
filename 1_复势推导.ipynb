# 1 项目背景与核心问题

## 1.1 工程背景
在理想流体力学中，圆柱绕流是平面势流的经典问题。单纯均匀来流复势（$U_0$ 为来流速度）无法满足**流体不能穿透圆柱表面（法向速度为零）**的物理边界条件。

本阶段目标是：通过引入偶极子项，构造一个既符合物理边界条件、又保持解析性的复势函数，为后续流场可视化与工程分析奠定数学基础。

## 1.2 基本假设
本推导基于理想流体平面势流假设：

- 不可压缩流体：密度 $\rho$ 为常数，连续性方程 $\nabla\cdot\boldsymbol{v}=0$；
- 无旋流动：速度场存在速度势函数 $\varphi$，满足 $\boldsymbol{v}=\nabla\varphi$；
- 平面流动：流动仅在 $x\text{-}y$ 平面内，所有物理量与 $z$ 坐标无关；
- 边界条件：圆柱表面 $r=a$ 处法向速度为 $0$，即流线贴合圆柱表面。

# 2 复势函数构造思路

## 2.1 基础流场：均匀来流
沿 $x$ 轴正方向的均匀来流复势为：
$$
F_1(z)=U_0 z
$$
其中 $z=x+iy$ 为复变量，$U_0$ 为来流速度。

速度势：
$$
\varphi_1=U_0 r\cos\theta
$$

流函数：
$$
\psi_1=U_0 r\sin\theta
$$

该流场的流线为平行于 $x$ 轴的直线，无法满足圆柱表面的闭合流线要求。

## 2.2 引入偶极子项
为满足圆柱边界的流线闭合条件，在均匀来流基础上引入偶极子复势：
$$
F_2(z)=\frac{M}{z}
$$
其中 $M$ 为偶极子强度，其物理本质为源-汇对极限（源与汇无限靠近且强度同步增大），可改变流场拓扑结构，使流线在圆柱表面闭合。

速度势：
$$
\varphi_2=\frac{M}{r}\cos\theta
$$

流函数：
$$
\psi_2=-\frac{M}{r}\sin\theta

# 3 边界条件匹配与偶极子强度推导

## 3.1 边界条件：圆柱表面法向速度为零
圆柱表面 $r=a$ 处，流体不可穿透，要求径向速度 $v_r=0$。

径向速度由速度势的偏导数定义：
$$
v_r=\frac{\partial\varphi}{\partial r}
$$

总复势为均匀来流与偶极子的叠加：
$$
F(z)=U_0 z+\frac{M}{z}
$$

对应总速度势：
$$
\varphi=U_0 r\cos\theta+\frac{M}{r}\cos\theta
$$

## 3.2 推导偶极子强度 $M$
对 $\varphi$ 求径向偏导，得到径向速度：
$$
v_r=\left(U_0-\frac{M}{r^2}\right)\cos\theta
$$

代入圆柱表面边界条件 $r=a$ 且 $v_r=0$：
$$
\left(U_0-\frac{M}{a^2}\right)\cos\theta=0
$$

对任意 $\theta$ 成立，因此系数必须满足：
$$
U_0-\frac{M}{a^2}=0
$$

解得偶极子强度：
$$
M=U_0 a^2

# 4 合成复势与极坐标展开

## 4.1 合成复势表达式
将 $M=U_0 a^2$ 代入偶极子复势，与均匀来流复势叠加，得到圆柱绕流总复势：
$$
F(z)=U_0\left(z+\frac{a^2}{z}\right)
$$
其中 $a$ 为圆柱半径，该复势在 $|z|\ge a$ 区域内解析，满足理想流体平面势流的数学要求。

## 4.2 极坐标形式展开
将 $z=re^{i\theta}$ 代入总复势，分离实部（速度势）与虚部（流函数）：

速度势：
$$
\varphi=U_0\left(r+\frac{a^2}{r}\right)\cos\theta
$$

流函数：
$$
\psi=U_0\left(r-\frac{a^2}{r}\right)\sin\theta
$$

## 4.3 边界条件验证
在圆柱表面 $r=a$ 处：
- 流函数：$\psi=0$，即圆柱表面为一条流线，满足流线闭合条件；
- 径向速度：$v_r=0$，满足无穿透边界条件。

# 5 极坐标柯西-黎曼（C-R）方程验证

## 5.1 极坐标C-R方程形式
解析复势需满足极坐标C-R方程：
$$
\begin{cases}
\dfrac{\partial\varphi}{\partial r}=\dfrac{1}{r}\dfrac{\partial\psi}{\partial\theta}\\[6pt]
\dfrac{1}{r}\dfrac{\partial\varphi}{\partial\theta}=-\dfrac{\partial\psi}{\partial r}
\end{cases}
$$

## 5.2 代入验证
第一式验证：
$$
\text{左侧：} \quad \dfrac{\partial\varphi}{\partial r}=U_0\left(1-\dfrac{a^2}{r^2}\right)\cos\theta
$$
$$
\text{右侧：} \quad \dfrac{1}{r}\dfrac{\partial\psi}{\partial\theta}=U_0\left(1-\dfrac{a^2}{r^2}\right)\cos\theta
$$
结论：左右两侧相等，第一式成立。

第二式验证：
$$
\text{左侧：} \quad \dfrac{1}{r}\dfrac{\partial\varphi}{\partial\theta}=-U_0\left(1+\dfrac{a^2}{r^2}\right)\dfrac{\sin\theta}{r}
$$
$$
\text{右侧：} \quad -\dfrac{\partial\psi}{\partial r}=-U_0\left(1+\dfrac{a^2}{r^2}\right)\dfrac{\sin\theta}{r}
$$
结论：左右两侧相等，第二式成立。

## 5.3 验证结论
圆柱绕流复势 $F(z)$ 在 $r\ge a$ 区域内满足极坐标C-R方程，证明其为解析函数，符合理想流体平面势流的数学要求。

# 6 复速度与物理场分析

## 6.1 复速度定义
复速度为复势的导数：
$$
\overline{V}=\frac{dF}{dz}=U_0\left(1-\frac{a^2}{z^2}\right)
$$

极坐标下速度分量：
径向速度：
$$
v_r=U_0\left(1-\frac{a^2}{r^2}\right)\cos\theta
$$

周向速度：
$$
v_\theta=-U_0\left(1+\frac{a^2}{r^2}\right)\sin\theta
$$

## 6.2 圆柱表面物理特性
在 $r=a$ 处：
- 径向速度 $v_r=0$，满足无穿透边界条件；
- 周向速度 $v_\theta=-2U_0\sin\theta$，合速度大小 $|V|=2U_0|\sin\theta|$；
- 压力系数（由伯努利方程）：$C_p=1-4\sin^2\theta$，体现圆柱表面压力分布规律。

## 7.2 复势函数构造
　　基于均匀来流与偶极子叠加原理，构造圆柱绕流复势模型。采用类封装实现参数管理与函数调用，默认参数：来流速度 \(U_0=1.0\)，圆柱半径 \(a=1.0\)。

````python
class FlowSimulator:
    """圆柱绕流复势模拟器
    默认参数 U=1.0, a=1.0
    """
    def __init__(self, U=1.0, a=1.0):
        self.U = U
        self.a = a

    def complex_potential(self, z):
        """计算圆柱绕流总复势"""
        return self.U * (z + self.a ** 2 / z)

    def get_doublet_strength(self):
        """计算满足边界条件的偶极子强度"""
        return 2 * np.pi * self.U * self.a ** 2
````

## 7.3 解析性验证
　　通过极坐标形式柯西-黎曼方程验证复势解析性，分别实现**符号推导验证**与**随机点数值残差校验**，双重验证解析性。

````python
def verify_cauchy_riemann_polar(r, theta, U, a):
    """极坐标C-R方程符号验证
    输入：符号变量
    输出：C-R方程残差（理想值为0）
    """
    phi = U * (r + a ** 2 / r) * sp.cos(theta)
    psi = U * (r - a ** 2 / r) * sp.sin(theta)
    
    cr1 = sp.simplify(sp.diff(phi, r) - (1 / r) * sp.diff(psi, theta))
    cr2 = sp.simplify((1 / r) * sp.diff(phi, theta) + sp.diff(psi, r))
    return cr1, cr2

def cr_residual_check(U=1.0, a=1.0, n_points=100):
    """C-R方程数值验证（随机点校验）
    输出：最大残差值（接近0表示验证通过）
    """
    r = np.random.uniform(a + 0.1, 5 * a, n_points)
    theta = np.random.uniform(0, 2 * np.pi, n_points)

    dphi_dr = U * (1 - a ** 2 / r ** 2) * np.cos(theta)
    dpsi_dtheta = U * (r - a ** 2 / r) * np.cos(theta)
    cr1 = np.max(np.abs(dphi_dr - (1 / r) * dpsi_dtheta))

    dphi_dtheta = -U * (r + a ** 2 / r) * np.sin(theta)
    dpsi_dr = U * (1 + a ** 2 / r ** 2) * np.sin(theta)
    cr2 = np.max(np.abs((1 / r) * dphi_dtheta + dpsi_dr))
    return cr1, cr2
````

## 7.4 运行与结果验证
　　使用默认参数进行复势计算与解析性验证，输出核心结果与验证信息。

````python
if __name__ == "__main__":
    sim = FlowSimulator()

    print("\n========== 任务1.1 复势构造结果 ==========")
    z_test = 2 + 1j
    print(f"测试点 z = {z_test}")
    print(f"复势 F(z) = {sim.complex_potential(z_test):.4f}")
    print(f"偶极子强度 κ = {sim.get_doublet_strength():.4f}")

    print("\n========== 任务1.2 符号解析性验证 ==========")
    r_s, t_s, U_s, a_s = sp.symbols('r theta U a', real=True, positive=True)
    cr1, cr2 = verify_cauchy_riemann_polar(r_s, t_s, U_s, a_s)
    print(f"C-R 方程 1 残差: {cr1}")
    print(f"C-R 方程 2 残差: {cr2}")

    print("\n========== 任务1.2 数值验证 ==========")
    max1, max2 = cr_residual_check()
    print(f"C-R 方程 1 最大残差: {max1:.2e}")
    print(f"C-R 方程 2 最大残差: {max2:.2e}")
    print("✅ 极坐标C-R方程满足，复势解析性验证通过！")
````
